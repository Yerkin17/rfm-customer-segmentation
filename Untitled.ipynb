```python
import pandas as pd
df = pd.read_csv("online_retail.csv", encoding="unicode_escape")
df = df.dropna(subset=["CustomerID"])
df = df[df["Quantity"] > 0]
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["TotalPrice"] = df["Quantity"] * df["UnitPrice"]
print(df.info())
print(df.head())
```

```python
import datetime as dt

# 1. Задаем «текущую» дату. Так как данные заканчиваются в декабре 2011 года, 
# возьмем дату на 1 день позже максимальной в датасете
snapshot_date = df["InvoiceDate"].max() + dt.timedelta(days=1)

# 2. Считаем RFM для каждого клиента
rfm = df.groupby("CustomerID").agg(
    {
        "InvoiceDate": lambda x: (snapshot_date - x.max()).days,  # Recency
        "InvoiceNo": "nunique",  # Frequency
        "TotalPrice": "sum",  # Monetary
    }
)

# 3. Переименовываем колонки для удобства
rfm.columns = ["Recency", "Frequency", "Monetary"]

# Проверяем результат (первые 5 строк)
print(rfm.head())
```

```python
# Присваиваем ранги от 1 до 5 с помощью квантилей (qcut)
# Для Recency: меньшее значение лучше (поэтому ascending=False)
rfm["R_Score"] = pd.qcut(rfm["Recency"], 5, labels=[5, 4, 3, 2, 1])

# Для Frequency и Monetary: большее значение лучше (ascending=True)
# Обрати внимание на drop_duplicates=False на случай дубликатов в квантилях
rfm["F_Score"] = pd.qcut(
    rfm["Frequency"].rank(method="first"), 5, labels=[1, 2, 3, 4, 5]
)
rfm["M_Score"] = pd.qcut(rfm["Monetary"], 5, labels=[1, 2, 3, 4, 5])

# Склеиваем баллы в единый RFM-индекс (строку, например, "555" или "123")
rfm["RFM_Score"] = (
    rfm["R_Score"].astype(str)
    + rfm["F_Score"].astype(str)
    + rfm["M_Score"].astype(str)
)

# Проверяем результат
print(rfm.head())
```

```python
# Функция для присвоения текстовых сегментов на основе R и F оценок
def rfm_segment(df):
    if df["R_Score"] >= 4 and df["F_Score"] >= 4:
        return "Champions"  # Покупают часто, недавно и много тратящих
    elif df["R_Score"] >= 3 and df["F_Score"] >= 3:
        return "Loyal Customers"  # Лояльные клиенты
    elif df["R_Score"] >= 3 and df["F_Score"] <= 2:
        return "Potential Loyalist"  # Потенциальные лояльные
    elif df["R_Score"] <= 2 and df["F_Score"] >= 4:
        return "At Risk"  # Давно не были, но раньше покупали часто
    elif df["R_Score"] <= 2 and df["F_Score"] <= 2:
        return "Hibernating / Lost"  # Уснувшие / ушедшие
    else:
        return "Needs Attention"  # Требуют внимания


# Применяем функцию к каждому клиенту
rfm["Segment"] = rfm.apply(rfm_segment, axis=1)

# Считаем, сколько клиентов в каждом сегменте
segment_counts = rfm["Segment"].value_counts()
print(segment_counts)

# Проверяем итоговую табличку
print(rfm.head())
```
